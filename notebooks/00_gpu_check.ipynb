```python
import torch

if torch.cuda.is_available():
    print("✅ Success! PyTorch can see your GPU.")
    print(f"Device Name: {torch.cuda.get_device_name(0)}")
    print(f"CUDA Version PyTorch was built with: {torch.version.cuda}")
else:
    print("❌ Error: PyTorch CANNOT see your GPU.")
    print("This means your model will run on the CPU (which will be extremely slow).")
    print("Please check your NVIDIA drivers and PyTorch CUDA installation.")
```
